```python
from skimage.data import astronaut
from skimage.segmentation import slic, mark_boundaries
from skimage.color import rgb2lab
import matplotlib.pyplot as plt
```

```python
from deephierarchy.graph import build_rag, weight_rag_weighted_area_dist
from deephierarchy.core import C4
from deephierarchy.labeling import *
from deephierarchy.hierarchy import bpt
```

```python
img = astronaut()
```

```python
plt.figure(figsize=(8, 8))
plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
label_map = slic(img, start_label=0)
```

```python
plt.figure(figsize=(8, 8))
plt.imshow(mark_boundaries(img, label_map))
plt.axis("off")
plt.show()
```

```python
rag = build_rag(label_map, C4)
```

```python
area = compute_area(label_map)
sum_v = compute_sum(label_map, rgb2lab(img))
```

```python
rag_w = weight_rag_weighted_area_dist(rag, area, sum_v)
```

```python
t = bpt(rag, rag_w, area, sum_v, label_map)
```
